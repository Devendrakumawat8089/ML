```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic.csv")
```

```python
df.head()
```

```python
df['Age'].describe()
```

```python
sns.boxplot( x = df['Age'])
```

```python
# Finding Percentile
```

```python
percentile25 = df['Age'].quantile(0.25)
percentile25
```

```python
percentile75 = df['Age'].quantile(0.75)
percentile75
```

```python
IQR =   percentile75 - percentile25
IQR
```

```python
upper_limit = percentile75 + 1.5*IQR
upper_limit
```

```python
lower_limit = percentile25 - 1.5*IQR
lower_limit
```

```python
# Finding our outlier
```

```python
df[df['Age'] < upper_limit ]
```

```python
df[df['Age'] < lower_limit ]
```

```python
# Trimming (Outlier Removing technique 1)

newdf = df[df['Age'] < upper_limit]
```

```python
newdf
```

```python
# comparision
```

```python
plt.figure(figsize=(15,5))

plt.subplot(222)
sns.boxplot(x=df['Age'])


plt.figure(figsize=(15,5))

plt.subplot(222)
sns.boxplot(x=newdf['Age'])
```

```python
# Capping (Outlier Removing technique 2)
```

```python
new_df_cap = df.copy()
```

```python
new_df_cap['Age'] = np.where(
    new_df_cap['Age'] > upper_limit,
    upper_limit,


    np.where(
        new_df_cap['Age'] < lower_limit,
        lower_limit,
        new_df_cap['Age']
    )
)
```

```python
# new_df_cap['Age'].max()
new_df_cap.head()
```

```python
plt.figure(figsize=(15,8))

plt.subplot(222)
sns.boxplot(x=df['Age'])


plt.figure(figsize=(15,5))

plt.subplot(222)
sns.boxplot(x=new_df_cap['Age'])
```
