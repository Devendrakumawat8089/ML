Task 3.1: Missing Value Treatment

For numerical columns:
Apply:
* Mean Imputation
* Median Imputation
For categorical columns:
Apply:

* Mode Imputation
* Unknown Category Imputation

Mean Imputation (Numerical  Columns)

```python
mean_df = df.copy()

num_cols = mean_df.select_dtypes(include=np.number).columns

mean_df[num_cols] = mean_df[num_cols].fillna(mean_df[num_cols].mean())

mean_df[num_cols]
```

Median Imputation (Numerical Columns)

```python
median_df = df.copy()

num_cols = median_df.select_dtypes(include=np.number).columns

median_df[num_cols] = median_df[num_cols].fillna(median_df[num_cols].median())

median_df[num_cols]
```

mode for Categorical Columns

```python
mode_df = df.copy()

cat_cols = mode_df.select_dtypes(include=['object', 'category']).columns

for col in cat_cols:
    mode_df[col] = mode_df[col].fillna(mode_df[col].mode()[0])
print(mode_df[col])
```

```python
unknown_df = df.copy()
unknown_df[cat_cols] = unknown_df[cat_cols].fillna('Unknown')
unknown_df[cat_cols]
```
