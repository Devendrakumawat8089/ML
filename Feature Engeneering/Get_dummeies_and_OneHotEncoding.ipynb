```python
import numpy as np
```

```python
data = {
    'Name':['Rahul','Ram','Shyam'],
    'City': ['Dehli','Jaipur','Bhratpur']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)
df
```

By Usign the Pandas get_dummies Method 

```python
get_dummies_01 = pd.get_dummies(df['City'] , dtype=int)
get_dummies_01
```

By Usign The Skite-Learn OneHotEncoding mehtod 

Step 1 : First We import OneHotEncoder from the Skite-Learn 

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
```

Step 2 : fit_transform :  That Means Lean And Apply 

```python
d = ohe.fit_transform(df[['City']])
```

```python
#  Step 3 :  Convert Into the DataFrmae 
```

get_feature_names_out() : Are used to view the columns name 

```python
# Convert Into The DataFrame
Ohe_DataFrmae = pd.DataFrame(d , columns= ohe.get_feature_names_out())
Ohe_DataFrmae
```
