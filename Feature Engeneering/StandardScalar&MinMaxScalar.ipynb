# Remove Outliers

```python
df = pd.DataFrame({
    'Salary':[500, 20000 , 25000 , 27000,29000,30000,31000,50000]
})

Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1 

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['remover_outliers'] = df['Salary'].clip(upper , lower)
```
