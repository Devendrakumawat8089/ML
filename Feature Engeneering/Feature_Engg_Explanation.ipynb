```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'colors':['Red','Green','Blue'],
    'Date':['2024-12-01','2023-11-04','2026-06-01']
})
df['clean_date'] = pd.to_datetime(df['Date'])
df['clean_date']

df['year'] = df['clean_date'].dt.year
df['month'] = df['clean_date'].dt.month
df['day'] = df['clean_date'].dt.day
df
```
