Task 5.1: IQR Method

For every numerical column:
Calculate:
* Q1
* Q3
* IQR
* Lower Bound
* Upper Bound

Identify:
* Number of outliers
Create:

* Boxplots before treatment
Questions

1. Which feature has the most outliers?
2. Why are outliers important?

```python
o = []
for col in num_cols:
    Q1 = df[col].quantile(0.25),
    Q3 = df[col].quantile(0.75)
    
    IQR = Q1 - Q3
    
    lower_bound = Q1 - (1.5 * IQR)
    
    Upper_Bound = Q3 - (1.5 * IQR)
    o.append({
        'Columns': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': Upper_Bound,
        'No. of Outliers': len(o)
    })
out_df = pd.DataFrame(o)

out_df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

for col in num_cols:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

Task 5.2: Z-Score Method

Calculate:
* Z-score for every numerical feature.
Identify observations where:
* |Z-score| > 3

Questions

1. Compare IQR and Z-score methods.
2. Which method performs better on skewed data?

```python
z_outliers = []

for col in num_cols:
    
    mean = df[col].mean()
    std = df[col].std()

    z_scores = (df[col] - mean) / std

    outliers = df[abs(z_scores) > 3]

    z_outliers.append({
        'Column': col,
        'Mean': mean,
        'Std Dev': std,
        'No. of Outliers': len(outliers)
    })

z_df = pd.DataFrame(z_outliers)

z_df
```

Q1. Compare IQR and Z-Score Methods

```python
# IQR is based on quartiles and is less affected by extreme values, whereas Z-score relies 
# on mean and standard deviation and is more sensitive to outliers.
```

Q2. Which Method Performs Better on Skewed Data?

```python
# IQR performs better on skewed data because it is less affected by extreme values.
```

Part 10: Skewness Analysis

Calculate skewness for all numerical features.
Apply:
* Log Transformation
* Square Root Transformation
* Compare skewness before and after.
Questions

1. Which transformation reduced skewness the most?
2. Why is skewness problematic?

```python
num_df = df.select_dtypes(include=['int64', 'float64'])

skew_before = num_df.skew()

skew_before.sort_values(ascending=False)
```

```python
import numpy as np

log_df = num_df.copy()

for col in num_df.columns:
    
    if (num_df[col] > 0).all():
        log_df[col] = np.log1p(num_df[col])

log_skew = log_df.skew()

log_skew
```

```python
sqrt_df = num_df.copy()

for col in num_df.columns:
    
    if (num_df[col] >= 0).all():
        sqrt_df[col] = np.sqrt(num_df[col])

sqrt_skew = sqrt_df.skew()

sqrt_skew
```

```python
comparison = pd.DataFrame({
    'Original Skewness': skew_before,
    'Log Transform': log_skew,
    'Sqrt Transform': sqrt_skew
})

comparison
```

```python
comparison_abs = pd.DataFrame({
    'Original': abs(skew_before),
    'Log': abs(log_skew),
    'Sqrt': abs(sqrt_skew)
})

comparison_abs
```

```python
# Before Transformation 
```

```python
import matplotlib.pyplot as plt

num_df.hist(figsize=(15,10))
plt.suptitle("Original Features")
plt.show()
```

```python
# After Log Transformation

log_df.hist(figsize=(15,10))
plt.suptitle("Log Transformed Features")
plt.show()
```

```python
# After Square Root Transformation

sqrt_df.hist(figsize=(15,10))
plt.suptitle("Square Root Transformed Features")
plt.show()
```

```python
# 1. Which transformation reduced skewness the most?
# Anser: Log Transformation generally reduced skewness the most because it compresses large values more aggressively.

# 2. Why is skewness problematic?
# Anser:  Skewness can distort statistical measures and reduce the performance of models that assume a normal distribution
```

```python
# Employees with fewer promotions are more likely to leave.
sns.boxplot(
    x='Attrition',
    y='YearsSinceLastPromotion',
    data=df
)

plt.show()
```

```python
# Younger employees leave more frequently.
sns.histplot(
    data=df,
    x='Age',
    hue='Attrition',
    kde=True
)

plt.show()
```

```python
# Lower job satisfaction leads to higher attrition.
sns.countplot(
    x='JobSatisfaction',
    hue='Attrition',
    data=df
)

plt.show()
```

```python
# Employees with fewer years at the company leave more often.
sns.boxplot(
    x='Attrition',
    y='YearsAtCompany',
    data=df
)

plt.show()
```
