**Step 9: Bivariate Analysis**


A. Numerical vs Numerical

```python
sns.scatterplot(
    x='GrLivArea',
    y='SalePrice',
    data=df
)

# This feature is important becuase the as the LivArea is increasing then the price is also increasing
# So, It means we to keep this feature
```

```python
df.corr(
    numeric_only=True
)['SalePrice'].sort_values(
    ascending=False
)

# From here we can select the important features from the data to train a model which will affect the model
# training..The data or the column is nearer to the 1 they're the important features(we have to keep that feature)
# Direct relation

# The columns which are around 0 they're the least important they doesn't show any of the relation..

# The columns which are in negative then they're the also the important features to the find the pattern..
# It means that they're indirectly correlated..
```
