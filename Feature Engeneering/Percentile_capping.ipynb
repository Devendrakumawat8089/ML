# Percentile Capping

```python
#  Percetile capping is the method to handle the outliers in which we cap the outliers wit their percentile value
# for left skewed data we use the 0.05 quantile value of the 5the percentile of the total values
# for right skewd data we use the 95th percentile value of the total for capping the outliews. 
```

```python
import pandas as pd
df = pd.DataFrame({
    'Salary':[500,20000 , 25000 ,27000 , 29000 , 30000 , 31000 , 50000]
})

lower = df['Salary'].quantile(0.10)
upper = df['Salary'].quantile(0.90)

df['percentile_capped'] = df['Salary'].clip(lower , upper)

df
```

```python
import matplotlib.pyplot as plt
```
