# Variance

Variance Is An Error Which Is Caused by Data Sensitivity (whenever the data varries the error increases)

# Bias

Bias Is And Error in which Model Is Too Simple TO UnderStand The Data Patterns That's Why The Model Will Not Perform Good On The Training As Well not for the Testing Also

# OverFitting

Overfitting in machine learning happens when a model memorizes the training data too closely—including its noise and random quirks—instead of learning generalizable patterns

# UnderFitting

Underfitting in machine learning occurs when a model is too simple to capture the underlying patterns in the data.

# bias-variance tradeoff

The bias-variance tradeoff is a fundamental concept in machine learning that governs a model's ability to generalize
