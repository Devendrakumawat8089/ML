Q1. What is Feature Engeneering ?.

Ans: in Feature Engeneeriung we basically create more features or reduce features on the basics of our requirnment. 

```python
import numpy as np
```

```python
data = {
    'Name':['Rahul','Ram','Shyam'],
    'City': ['Dehli','Jaipur','Bhratpur']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)
df
```

By Usign the Pandas get_dummies Method 

```python
get_dummies_01 = pd.get_dummies(df['City'] , dtype=int)
get_dummies_01
```

By Usign The Skite-Learn OneHotEncoding mehtod 

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
```

```python
d = ohe.fit_transform(df[['City']] , columns = ohe.get_feature_names_out())
```
