```python
import pandas as pd
import numpy as np
```

```python
from sklearn.linear_model import Ridge , Lasso , ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error , r2_score
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("insurance.csv")
df.head(2)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb = LabelEncoder()
```

```python
df['sex'] = lb.fit_transform(df['sex'])
df['smoker'] = lb.fit_transform(df['smoker'])
df['region'] = lb.fit_transform(df['region'])
```

```python
df.head()
```

```python
x = df.drop(columns=['charges'])
y=df['charges']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2 , random_state=42)
```

```python
scalar = StandardScaler()
```

```python
x_train = scalar.fit_transform(x_train)
x_test = scalar.fit_transform(x_test)
```

```python
# ## Ridge:::::
# Ridge adds l2 Penatly.
# It reduces Coefficient values but does not make them Exactly
```

```python
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(x_train, y_train)
```

```python
ridge_pred = ridge_model.predict(x_test)
```

```python
ridge_mse = mean_squared_error(y_test , ridge_pred)
ridge_r2 = r2_score(y_test , ridge_pred)
print('MSE: ' ,ridge_mse)
print("R2_Score: ",ridge_r2)
print("\n Cofficients : " , ridge_model.coef_)
```

```python
# Lasso adds L1 Penalty 
# It Can Make Some Coefficients Exactly Zero To Performs Feature Selection 
```

```python
lasso_model = Lasso(alpha=0.1)
lasso_model.fit(x_train , y_train)
```

```python
lasso_pred = lasso_model.predict(x_test)
```

```python
lasso_mse = mean_squared_error(y_test , lasso_pred)
lasso_r2 = r2_score(y_test , lasso_pred)

print("Mse: ", lasso_mse)
print("r2_score: ", lasso_r2)
print("\n Cofficient " , lasso_model.coef_)
```

```python
# ## ElasticNet combines both:
# l1 Regularization (Lasso)
# l2 Regularization (Ridge)
```

```python
elasticnet_model = ElasticNet(alpha=0.1 , l1_ratio=0.5)
elasticnet_model.fit(x_train , y_train)
```

```python
elasticnet_pred = elasticnet_model.predict(x_test)
```

```python
elasticnet_mse = mean_squared_error(y_test , elasticnet_pred)
elasticnet_r2 = r2_score(y_test , elasticnet_pred)

print("Mse: " ,elasticnet_mse)
print("r2_score: ", elasticnet_r2)

print("\n Conficient : ", elasticnet_model.coef_)
```

```python
# Cross Validation
```

```python
from sklearn.linear_model import LassoCV
lasso_cv= LassoCV(
    alphas=[0.001 , 0.01 , 0.1 , 1 , 10],
    cv=5,
    random_state=42
)

lasso_cv.fit(x_train ,y_train)
print("Best Alpha : " , lasso_cv.alpha_)
```

```python
from sklearn.linear_model import ElasticNetCV

elasticnet_cv = ElasticNetCV(
    l1_ratio=[0.1 , 0.3 , 0.5 , 0.7 , 0.9 , 1],
    alphas=[0.001 , 0.01 , 0.1 , 1 ,10],
    cv = 5,
    random_state=42
    
)

elasticnet_cv.fit(x_train , y_train)
print("Best Alpha: ", elasticnet_cv.alpha_)
print("Best l1_ratio: ", elasticnet_cv.l1_ratio_)
```
