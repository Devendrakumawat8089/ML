# Ordinal 

Ordinal Encoding Me Hme Categories ke bases pr ,, Encode Kr Skhte Hai 

```python
data  = {
    'Pass/Fali':['Pass','Fail' , 'Pass'],
    'Customer SatisFaction': ['Good' , 'Better' , 'Best']
}
```

```python
df = pd.DataFrame(data)
df
```

```python
oe = OrdinalEncoder(categories=[['Good' , 'Best' , 'Better']])
```

```python
df['Encoded_Data'] = oe.fit_transform(df[['Customer SatisFaction']])
df['Encoded_Data']
```
