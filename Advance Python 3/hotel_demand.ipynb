Part 6: Outlier Detection
Task 6.1: IQR Method
For every numerical feature:
# Calculate:
* Q1
* Q3
* IQR
* Lower Bound
* Upper Bound
# Identify:
* Number of Outliers
* Create:
* Boxplots

```python
l =[]
for col in num_col:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3-Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)
    
    l.append({
        'Columns': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'No. of Outliers': len(l)
    })
    
    out_df = pd.DataFrame(l)
    
out_df
```

```python
n = out_df.select_dtypes(include=['int64','float64']).columns
```

Task 6.2: Z-Score Method
Calculate:
* Z-score
Identify:
* |Z| > 3

```python
z_outliers = []

for col in num_col:
    
    mean = df[col].mean()
    std = df[col].std()

    z_scores = (df[col] - mean) / std

    outliers = df[abs(z_scores) > 3]

    z_outliers.append({
        'Column': col,
        'Mean': mean,
        'Std Dev': std,
        'No. of Outliers': len(outliers)
    })

z_df = pd.DataFrame(z_outliers)

z_df
```

Part 7: Outlier Treatment
# Apply:
# Method 1
* Remove Outliers
# Method 2
* Capping
# Method 3
* Winsorizatio

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in num_col:

    plt.figure(figsize=(8,2))

    plt.subplot(1,2,1)
    sns.boxplot(x=df[col])
    plt.title(f'Before - {col}')

    plt.subplot(1,2,2)
    # sns.boxplot(x=df_remove[col])
    plt.title(f'After Removal - {col}')

    plt.tight_layout()
    plt.show()
```

```python
df_cap = df.copy()

for col in num_col:
    Q1 = df_cap[col].quantile(0.25)
    Q3 = df_cap[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_cap[col] = df_cap[col].clip(lower, upper)

df_cap.head()
```

```python
from scipy.stats.mstats import winsorize

df_win = df.copy()

for col in num_col:
    df_win[col] = winsorize(df_win[col], limits=[0.05, 0.05])

df_win.head()
```
